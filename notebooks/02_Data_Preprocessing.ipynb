# Data Preprocessing

## Intelligent Heart Disease Risk Prediction and Medical Decision Support System

This notebook focuses on cleaning and preprocessing the UCI Heart Disease dataset before machine learning model development.

### Main Objectives

- Load the original heart disease dataset.
- Inspect the data before preprocessing.
- Handle missing values.
- Prepare the target variable for binary classification.
- Prepare the features for machine learning.
- Export the cleaned dataset for subsequent modeling.

## 1. Import Libraries

The required Python libraries for data preprocessing are imported in this section.

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully.")
```

## 2. Load Raw Dataset

The original UCI Heart Disease dataset is loaded from the raw CSV file for preprocessing.

```python
# Load the raw dataset from Google Drive

file_path = '/content/drive/MyDrive/Heart_Disease_Project/data/heart_disease_raw.csv'

df = pd.read_csv(file_path)

print("Raw dataset loaded successfully.")
```

```python
import os

print("Current folder:")
print(os.getcwd())

print("\nFiles in current folder:")
print(os.listdir())
```

## 3. Inspect Data Before Preprocessing

This section verifies the structure and quality of the raw dataset before applying any preprocessing steps.

```python
# Display the first five rows
display(df.head())

# Display dataset shape
print("Dataset shape:", df.shape)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())
```

## 4. Handle Missing Values

Missing values were identified in the `ca` and `thal` features.  
Since these variables contain discrete categorical/ordinal codes, missing values will be imputed using the mode (most frequent value) of each feature.

```python
# Check the most frequent value (mode) for columns with missing values

print("Mode of ca:", df["ca"].mode()[0])
print("Mode of thal:", df["thal"].mode()[0])
```

```python
# Fill missing values using the mode

df["ca"] = df["ca"].fillna(df["ca"].mode()[0])
df["thal"] = df["thal"].fillna(df["thal"].mode()[0])

print("Missing values handled successfully.")
```

```python
# Verify missing values after imputation

print("Missing values after preprocessing:")
print(df.isnull().sum())
```

## 5. Check Duplicate Records

Duplicate records are checked before further preprocessing to ensure data quality.

```python
# Check for duplicate rows

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

## 6. Prepare Binary Target Variable

The original `num` target contains values from 0 to 4.  
For this project, the prediction task is formulated as binary classification:

- `0` = No Heart Disease
- `1` = Heart Disease

All original target values greater than 0 are converted to 1.

```python
# Convert the original target into binary classification

df["target"] = (df["num"] > 0).astype(int)

# Remove the original target column
df = df.drop(columns=["num"])

print("Binary target created successfully.")
```

```python
# Verify the binary target

print("Target values:")
print(df["target"].unique())

print("\nTarget distribution:")
print(df["target"].value_counts().sort_index())

print("\nDataset shape:", df.shape)
```

## 7. Separate Features and Target

The cleaned dataset is separated into input features (`X`) and the binary target variable (`y`) in preparation for machine learning.

```python
# Separate input features and target

X = df.drop(columns=["target"])
y = df["target"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

## 8. Final Data Quality Check

A final inspection is performed to verify that the dataset is clean and ready for export and subsequent machine learning tasks.

```python
# Final data quality check

print("Dataset shape:", df.shape)
print("\nTotal missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())

print("\nTarget distribution:")
print(df["target"].value_counts().sort_index())

print("\nData types:")
print(df.dtypes)
```

## 9. Export Cleaned Dataset

The cleaned and preprocessed dataset is exported as a CSV file for use in the machine learning modeling stage.

```python
# Export the cleaned dataset

df.to_csv("heart_disease_cleaned.csv", index=False)

print("Cleaned dataset saved successfully.")
```
